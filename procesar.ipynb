#  Procesamiento

## Imports y rutas

```python
import pandas as pd
from pathlib import Path

# Carpetas
entrada = Path("datos/originales")
salida = Path("datos/procesados")

salida.mkdir(parents=True, exist_ok=True)

archivos = sorted(entrada.glob("*.xlsx"))

print(f"Archivos encontrados: {len(archivos)}")
```

# Seleccion de columnas para el procesamiento de los datos


```python
columnas = [
    "ANO", "SEMANA", "COD_DPTO_O", "COD_MUN_O",
    "Departamento_ocurrencia", "Municipio_ocurrencia",
    "AJUSTE", "PAC_HOS", "SEXO", "EDAD", "UNI_MED",
    "AREA", "TIP_SS", "PER_ETN", "GP_DESPLAZ",
    "GP_MIGRANT", "GP_GESTAN", "estrato",
    "TIP_CAS", "nom_est_f_caso"
]

lista_df = []

for archivo in archivos:

    print(f"Leyendo: {archivo.name}")

    df = pd.read_excel(
        archivo,
        usecols=lambda c: c in columnas
    )

    # Crear columnas faltantes
    for col in columnas:
        if col not in df.columns:
            df[col] = pd.NA

    # Filtro de casos válidos
    df["AJUSTE"] = df["AJUSTE"].astype("string").str.strip()

    df = df[df["AJUSTE"].isin(["0", "3", "4", "5"])].copy()

    lista_df.append(df)

df_consolidado = pd.concat(
    lista_df,
    ignore_index=True
)

print(f"\nTotal de registros: {len(df_consolidado):,}")
print(f"Total de variables: {df_consolidado.shape[1]}")
```

# Guardar datos procesados en un archivo parquet
    

```python
ruta_parquet = salida / "dengue_consolidado.parquet"

df_consolidado.to_parquet(
    ruta_parquet,
    index=False
)

print("\n¡Parquet creado correctamente!")
print(f"Ruta: {ruta_parquet.resolve()}")
print(f"Registros: {len(df_consolidado):,}")
```
